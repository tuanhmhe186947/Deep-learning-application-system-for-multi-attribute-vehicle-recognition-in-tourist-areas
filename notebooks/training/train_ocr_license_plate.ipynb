```python
VALID_PROVINCES = [str(i) for i in range(11, 100)] + ['80']

# Mapping Chá»¯ -> Sá»‘
DICT_CHAR_TO_NUM = {
    'Z': '2', 'B': '8', 'D': '0', 'O': '0', 'Q': '0',
    'S': '5', 'G': '6', 'I': '1', 'L': '1', 'A': '4',
    'J': '1', 'T': '1'
}
# Mapping Sá»‘ -> Chá»¯
DICT_NUM_TO_CHAR = {
    '2': 'Z', '8': 'B', '0': 'D', '5': 'S', '6': 'G',
    '1': 'I', '4': 'A', '7': 'T'
}

def standardize_plate_chars(char_list):
    """
    Nháº­n vĂ o list kĂ½ tá»± thĂ´ -> Tráº£ vá» chuá»—i Ä‘Ă£ sá»­a lá»—i logic.
    DĂ¹ng cho cáº£ Prediction vĂ  Ground Truth.
    """
    final_chars = []

    for i, char in enumerate(char_list):
        # Vá»‹ trĂ­ 0, 1 (MĂ£ tá»‰nh) -> Pháº£i lĂ  Sá»
        if i == 0 or i == 1:
            if char in DICT_CHAR_TO_NUM:
                char = DICT_CHAR_TO_NUM[char]

        # Vá»‹ trĂ­ 2 (Series) -> Pháº£i lĂ  CHá»®
        elif i == 2:
            if char in DICT_NUM_TO_CHAR:
                char = DICT_NUM_TO_CHAR[char]

        # Vá»‹ trĂ­ > 2 (Sá»‘ Ä‘uĂ´i) -> Pháº£i lĂ  Sá»
        elif i > 2:
            if char in DICT_CHAR_TO_NUM:
                char = DICT_CHAR_TO_NUM[char]

        final_chars.append(char)

    return "".join(final_chars)

def advanced_ocr_logic(results_obj, class_list, threshold=0.25):
    """
    Xá»­ lĂ½ káº¿t quáº£ YOLOv8: Lá»c nhiá»…u -> Sáº¯p xáº¿p -> Chuáº©n hĂ³a -> Kiá»ƒm tra.
    LuĂ´n tráº£ vá» 2 giĂ¡ trá»‹: (text_result, warning_note)
    """
    raw_candidates = []

    for box in results_obj.boxes:
        if float(box.conf[0]) < threshold: continue

        coords = box.xyxy[0].tolist()
        w = coords[2] - coords[0]
        h = coords[3] - coords[1]
        if w > h * 1.5: continue

        raw_candidates.append({
            'char': class_list[int(box.cls[0])],
            'x': coords[0], 'y': coords[1], 'w': w, 'h': h,
            'conf': float(box.conf[0])
        })

    # Quan trá»ng: Tráº£ vá» 2 giĂ¡ trá»‹ Ä‘á»ƒ trĂ¡nh lá»—i "too many values to unpack"
    if not raw_candidates: return "", "No text found"

    heights = [item['h'] for item in raw_candidates]
    median_h = np.median(heights)

    # Loáº¡i bá» box nhá» hÆ¡n 50% chiá»u cao trung bĂ¬nh (á»‘c vĂ­t, dáº¥u cháº¥m)
    valid_boxes = [b for b in raw_candidates if b['h'] > median_h * 0.5]

    if not valid_boxes: return "", "All boxes filtered"

    valid_boxes.sort(key=lambda b: b['y']) # Sort theo Y
    rows = []
    current_row = [valid_boxes[0]]

    for i in range(1, len(valid_boxes)):
        box = valid_boxes[i]
        prev_box = current_row[-1]

        # PhĂ¢n dĂ²ng: Náº¿u chĂªnh lá»‡ch Y > 50% chiá»u cao -> DĂ²ng má»›i
        if (box['y'] - prev_box['y']) > (prev_box['h'] * 0.5):
            rows.append(current_row)
            current_row = [box]
        else:
            current_row.append(box)
    rows.append(current_row)

    # LĂ m pháº³ng list
    sorted_chars_list = []
    for row in rows:
        row.sort(key=lambda b: b['x'])
        sorted_chars_list.extend([item['char'] for item in row])

    result_string = standardize_plate_chars(sorted_chars_list)

    notes = []
    # Check mĂ£ tá»‰nh
    if "".join(result_string[:2]) not in VALID_PROVINCES:
        notes.append("Check MĂ£ Tá»‰nh")
    # Check Format (Regex)
    if not re.match(r"^\d{2}[A-Z]\d{4,5}$", result_string):
        notes.append("Check Format")

    return result_string, ", ".join(notes)

print("ÄĂ£ cáº­p nháº­t xong cĂ¡c hĂ m xá»­ lĂ½ Logic!")
```

```python
def get_gt_text(label_path):
    if not os.path.exists(label_path): return ""
    boxes = []
    with open(label_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            # cls, x, y, h
            boxes.append((int(parts[0]), float(parts[1]), float(parts[2]), float(parts[4])))

    if not boxes: return ""

    # Sáº¯p xáº¿p GT y há»‡t Prediction
    boxes.sort(key=lambda b: b[2]) # Sort Y
    rows = []
    curr = [boxes[0]]
    for i in range(1, len(boxes)):
        if (boxes[i][2] - curr[-1][2]) > (curr[-1][3] * 0.5):
             rows.append(curr); curr = [boxes[i]]
        else: curr.append(boxes[i])
    rows.append(curr)

    sorted_gt_chars = []
    for row in rows:
        row.sort(key=lambda b: b[1]) # Sort X
        sorted_gt_chars.extend([class_names[b[0]] for b in row])

    return standardize_plate_chars(sorted_gt_chars)
```

```python
def plot_smart_visual(img_path, model, pred_text, gt_text, cer):
    img = cv2.imread(img_path)
    if img is None: return

    # Predict láº¡i Ä‘á»ƒ láº¥y tá»a Ä‘á»™ váº½
    results = model(img, verbose=False)[0]
    img_draw = img.copy()

    # Cáº¥u hĂ¬nh váº½
    font = cv2.FONT_HERSHEY_SIMPLEX
    font_scale = 0.5; thickness = 1
    box_color = (0, 255, 0); text_bg = (0, 200, 0); text_col = (255, 255, 255)

    # Láº¥y data box & Lá»c
    boxes_data = []
    for box in results.boxes:
        if float(box.conf[0]) < conf_threshold: continue
        coords = box.xyxy[0].tolist()
        w, h = coords[2] - coords[0], coords[3] - coords[1]

        if w > h * 1.5: continue # Lá»c tá»· lá»‡

        boxes_data.append({
            'coords': [int(c) for c in coords],
            'char': class_names[int(box.cls[0])], # Láº¥y kĂ½ tá»± gá»‘c tá»« YOLO
            'center_x': int((coords[0] + coords[2]) / 2),
            'y': coords[1]
        })

    # Sáº¯p xáº¿p Ä‘á»ƒ Ă¡p dá»¥ng logic sá»­a lá»—i theo index
    boxes_data.sort(key=lambda x: x['center_x'])

    occupied_areas = []
    def is_intersect(r1, r2, pad=2):
        return not (r1[2]+pad <= r2[0] or r1[0]-pad >= r2[2] or r1[3]+pad <= r2[1] or r1[1]-pad >= r2[3])

    for i, item in enumerate(boxes_data):
        x1, y1, x2, y2 = item['coords']
        raw_char = item['char']

        display_char = raw_char
        if i == 0 or i == 1: # MĂ£ tá»‰nh -> Sá»‘
             if display_char in DICT_CHAR_TO_NUM: display_char = DICT_CHAR_TO_NUM[display_char]
        elif i == 2: # Series -> Chá»¯
             if display_char in DICT_NUM_TO_CHAR: display_char = DICT_NUM_TO_CHAR[display_char]
        elif i > 2: # Sá»‘ Ä‘uĂ´i -> Sá»‘
             if display_char in DICT_CHAR_TO_NUM: display_char = DICT_CHAR_TO_NUM[display_char]

        # Váº½ Box
        cv2.rectangle(img_draw, (x1, y1), (x2, y2), box_color, 1)

        # TĂ­nh vá»‹ trĂ­ nhĂ£n
        (tw, th), _ = cv2.getTextSize(display_char, font, font_scale, thickness)
        cx = int((x1 + x2)/2)
        lx1 = cx - tw//2; ly1 = y1 - th - 5
        lx2 = lx1 + tw;   ly2 = y1 - 5

        while True:
            collision = False
            curr = (lx1, ly1, lx2, ly2)
            for occ in occupied_areas:
                if is_intersect(curr, occ):
                    collision = True; break
            if collision:
                shift = th + 5; ly1 -= shift; ly2 -= shift
            else: break

        occupied_areas.append((lx1, ly1, lx2, ly2))

        if ly2 < y1 - 5:
            cv2.line(img_draw, (cx, ly2), (cx, y1), box_color, 1)

        # Váº½ nhĂ£n (Hiá»ƒn thá»‹ kĂ½ tá»± ÄĂƒ Sá»¬A Lá»–I)
        cv2.rectangle(img_draw, (lx1-2, ly1-2), (lx2+2, ly2+2), text_bg, -1)
        cv2.putText(img_draw, display_char, (lx1, ly2), font, font_scale, text_col, thickness)

    # Hiá»ƒn thá»‹ Matplotlib
    plt.figure(figsize=(10, 6))
    plt.imshow(cv2.cvtColor(img_draw, cv2.COLOR_BGR2RGB))
    title_color = 'red' if cer > 0 else 'green'
    plt.title(f"File: {os.path.basename(img_path)}\nGT: {gt_text} | Pred: {pred_text}\nCER: {cer:.2f}",
              color=title_color, fontsize=12, fontweight='bold')
    plt.axis('off')
    plt.show()

print("Báº¯t Ä‘áº§u quy trĂ¬nh OCR hĂ ng loáº¡t...")
image_files = glob.glob(os.path.join(val_images_path, "*.jpg")) + glob.glob(os.path.join(val_images_path, "*.png"))
ocr_results_data = []

for img_path in tqdm(image_files):
    base_name = os.path.basename(img_path)
    label_file = os.path.splitext(base_name)[0] + ".txt"
    label_path = os.path.join(val_labels_path, label_file)

    # 1. Láº¥y GT
    gt_text = get_gt_text(label_path)

    # 2. Predict YOLO
    results = model(img_path, verbose=False)[0]

    # 3. Xá»­ lĂ½ Logic OCR (ÄĂ£ qua hĂ m standardize_plate_chars bĂªn trong)
    pred_text, note = advanced_ocr_logic(results, class_names, conf_threshold)

    # 4. TĂ­nh CER
    if not gt_text and not pred_text: cer = 0.0
    elif not gt_text or not pred_text: cer = 1.0
    else: cer = jiwer.cer(gt_text, pred_text)

    ocr_results_data.append({
        'File Name': base_name,
        'Full Path': img_path,
        'Ground Truth': gt_text,
        'Prediction': pred_text,
        'CER Score': cer,
        'Tráº¡ng thĂ¡i': 'Sai' if cer > 0 else 'ÄĂºng',
        'Ghi chĂº': note
    })

df_ocr = pd.DataFrame(ocr_results_data)
mean_cer = df_ocr['CER Score'].mean()

print(f"\n" + "="*50)
print(f"Káº¾T QUáº¢ ÄĂNH GIĂ: Trung bĂ¬nh CER: {mean_cer:.4f}")
print("="*50)

# Lá»c áº£nh sai
df_errors = df_ocr[df_ocr['CER Score'] > 0].sort_values(by='CER Score', ascending=False)

if df_errors.empty:
    print("\nTuyá»‡t vá»i! CER = 0. Hiá»ƒn thá»‹ 3 áº£nh máº«u:")
    df_show = df_ocr.sample(min(3, len(df_ocr)))
else:
    print(f"\nTĂ¬m tháº¥y {len(df_errors)} áº£nh cĂ²n sai sĂ³t.")
    print("Hiá»ƒn thá»‹ TOP 5 áº£nh sai náº·ng nháº¥t Ä‘á»ƒ kiá»ƒm tra:")
    df_show = df_errors.head(5)

# Hiá»ƒn thá»‹ áº£nh
for _, row in df_show.iterrows():
    plot_smart_visual(
        img_path=row['Full Path'],
        model=model,
        pred_text=row['Prediction'],
        gt_text=row['Ground Truth'],
        cer=row['CER Score']
    )
```

```python
def predict_new_image(image_path, model):
    img = cv2.imread(image_path)
    if img is None:
        print(f"Lá»—i: KhĂ´ng tĂ¬m tháº¥y áº£nh táº¡i {image_path}")
        return

    results = model(img, verbose=False)[0]
    final_text, note = advanced_ocr_logic(results, class_names, conf_threshold)

    img_draw = img.copy()
    font = cv2.FONT_HERSHEY_SIMPLEX
    font_scale = 0.5
    thickness = 1
    box_color = (0, 255, 0)
    text_bg = (0, 200, 0)
    text_col = (255, 255, 255)

    boxes_data = []
    for box in results.boxes:
        if float(box.conf[0]) < conf_threshold: continue
        coords = box.xyxy[0].tolist()
        w = coords[2] - coords[0]
        h = coords[3] - coords[1]

        if w > h * 1.5: continue

        boxes_data.append({
            'coords': [int(c) for c in coords],
            'char': class_names[int(box.cls[0])],
            'center_x': int((coords[0] + coords[2]) / 2)
        })

    boxes_data.sort(key=lambda x: x['center_x'])

    occupied_areas = []
    def is_intersect(r1, r2, pad=2):
        return not (r1[2]+pad <= r2[0] or r1[0]-pad >= r2[2] or r1[3]+pad <= r2[1] or r1[1]-pad >= r2[3])

    for i, item in enumerate(boxes_data):
        x1, y1, x2, y2 = item['coords']
        display_char = item['char']

        if i == 0 or i == 1:
             if display_char in DICT_CHAR_TO_NUM: display_char = DICT_CHAR_TO_NUM[display_char]
        elif i == 2:
             if display_char in DICT_NUM_TO_CHAR: display_char = DICT_NUM_TO_CHAR[display_char]
        elif i > 2:
             if display_char in DICT_CHAR_TO_NUM: display_char = DICT_CHAR_TO_NUM[display_char]

        cv2.rectangle(img_draw, (x1, y1), (x2, y2), box_color, 2)

        (tw, th), _ = cv2.getTextSize(display_char, font, font_scale, thickness)
        cx = int((x1 + x2)/2)
        lx1 = cx - tw//2; ly1 = y1 - th - 5
        lx2 = lx1 + tw;   ly2 = y1 - 5

        while True:
            collision = False
            curr = (lx1, ly1, lx2, ly2)
            for occ in occupied_areas:
                if is_intersect(curr, occ):
                    collision = True; break
            if collision:
                shift = th + 5; ly1 -= shift; ly2 -= shift
            else: break
        occupied_areas.append((lx1, ly1, lx2, ly2))

        if ly2 < y1 - 5:
            cv2.line(img_draw, (cx, ly2), (cx, y1), box_color, 1)

        cv2.rectangle(img_draw, (lx1-2, ly1-2), (lx2+2, ly2+2), text_bg, -1)
        cv2.putText(img_draw, display_char, (lx1, ly2), font, font_scale, text_col, thickness)

    plt.figure(figsize=(10, 8))
    plt.imshow(cv2.cvtColor(img_draw, cv2.COLOR_BGR2RGB))

    plt.title(f"Káº¾T QUáº¢: {final_text}\n(Note: {note})", color='blue', fontsize=14, fontweight='bold')
    plt.axis('off')
    plt.show()

my_image_path = "/content/quy-dinh-bien-so-xe.jpg"
predict_new_image(my_image_path, model)
```
